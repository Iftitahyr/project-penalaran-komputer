```python
import pandas as pd
import re
import os
```

```python
input_path = '/content/drive/MyDrive/semester 6/Penalaran Komputer/UAS/CSV/putusan_ma__2025-06-23.csv'
df = pd.read_csv(input_path)
df.head()
```

```python
# Cek dan hapus duplikat berdasarkan nomor perkara (jika ada)
print("Jumlah awal:", len(df))
df = df.drop_duplicates(subset=['nomor'])
print("Jumlah setelah hapus duplikat berdasarkan nomor:", len(df))
```

```python
df = df[df['kata_kunci'].str.contains("pembunuhan", case=False, na=False)].copy()
print("Jumlah kasus pembunuhan:", len(df))
df.head()
```

```python
df['case_id'] = ['case_{:03d}'.format(i+1) for i in range(len(df))]
```

```python
def get_summary(text, n=3):
    if pd.isna(text):
        return ""
    sentences = re.split(r'(?<=[.!?]) +', text.strip())
    return ' '.join(sentences[:n])

df['ringkasan_fakta'] = df['text_pdf'].apply(lambda x: get_summary(x, n=3))
```

```python
def extract_pasal(text):
    if pd.isna(text):
        return ""
    match = re.findall(r'Pasal\s\d+[A-Za-z\s]*', text)
    return '; '.join(set(match)) if match else ""

df['pasal'] = df['text_pdf'].apply(extract_pasal)
```

```python
df['text_full'] = df['text_pdf']

```

```python
selected_cols = ['case_id', 'nomor', 'tanggal_register', 'tanggal_dibacakan', 'ringkasan_fakta', 'pasal', 'amar','catatan_amar', 'text_full']
df_out = df[selected_cols]
df_out.head()
```

```python
output_path = '/content/drive/MyDrive/semester 6/Penalaran Komputer/UAS/processed/cases.csv'
os.makedirs(os.path.dirname(output_path), exist_ok=True)
df_out.to_csv(output_path, index=False)
print("✅ File berhasil disimpan di:", output_path)
```
